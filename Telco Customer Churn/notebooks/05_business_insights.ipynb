# Customer Churn - Final Business Insights

## Executive Summary

This notebook presents the final business interpretation of the churn prediction analysis. 
It summarises key drivers of churn, model performance, and actionable retention strategies.

Customer churn rate remains approximately **26–27%**, representing a significant revenue risk.

### Two models were evaluated:

| Model               | Accuracy | Precision | Recall | ROC-AUC |
|---------------------|----------|-----------|--------|---------|
| Logistic Regression | 0.792    | 0.632     | 0.524  | 0.835   |
| Random Forest       | 0.789    | 0.626     | 0.511  | 0.824   |



### Key Takeaways

- Logistic Regression slightly outperformed Random Forest overall.
- ROC-AUC ≈ 0.83–0.84, indicating strong class separation.
- Recall ≈ 52%, meaning the model detects just over half of churners at the default threshold.

Logistic Regression is therefore selected as the prefered model due to:

- Slightly better recall
- Higher ROC-AUC
- Better interpretability

## Confusion Matrix Interpretation (Logistic Regression)

From the model’s confusion matrix:

- **True Negatives (TN):** 919  
- **False Positives (FP):** 114  
- **False Negatives (FN):** 178  
- **True Positives (TP):** 196  

### Business Meaning

- The model correctly identifies **196 churners**.
- It misses **178 churners**.
- Approximately **48% of churners remain undetected** at the default threshold.

This indicates that while overall accuracy is strong, recall can be improved.  
Threshold tuning could significantly increase churn detection performance.


## Key Statistical Drivers of Churn

From the correlation analysis:

### Strongest Relationships with Churn

- **Tenure vs Churn:** -0.35  
  → Strongest numeric relationship  
  → Short-tenure customers are highest risk  

- **MonthlyCharges vs Churn:** +0.19  
  → Higher pricing increases churn probability  

- **TotalCharges vs Churn:** -0.20  
  → Long-term cumulative customers are less likely to churn  



### Multicollinearity Observed

- **Tenure & TotalCharges:** 0.83  
- **MonthlyCharges & TotalCharges:** 0.65  

This suggests that TotalCharges is largely driven by tenure and monthly charges. High correlation between tenure and total charges suggests billing history largely reflects customer lifecycle stage.
  
Customer longevity (tenure) is the dominant driver of churn behaviour, highlighting early lifecycle engagement as a critical retention strategy.



## Feature Importance Interpretation (Random Forest)

### Clear Pattern Observed

Churn is primarily driven by:

- **Customer lifecycle stage (tenure)**
- **Pricing intensity**
- **Contract flexibility**
- **Payment behaviour**
- **Fibre service type**

Customers with shorter tenure, higher effective monthly spend, flexible contracts, and electronic check payment methods are significantly more likely to churn.

Two-year contracts appear strongly protective against churn.



### Strategic Business Insight

Churn risk is concentrated in early-lifecycle, high-cost, flexible-contract customers.  
Demographics (e.g., gender) show minimal predictive value.

Retention strategies should therefore focus on:
- Early customer engagement
- Pricing optimisation
- Incentivising long-term contracts
- Improving billing/payment experience

**Conclusion:** Behavioural and pricing variables dominate churn prediction, confirming that operational and contract design levers are more impactful than demographic targeting.

### 🎯 2. Contract Conversion Campaign

**Impact:**

### 🎯 4. Promote Automatic Payments

### 🎯 5. Price Sensitivity Monitoring

- Offer personalised retention pricing

Mitigates churn driven by pricing pressure.


## Model Performance — Business Framing

- Accuracy ≈ 73%

At the default threshold:

- ~52% of churners are correctly detected
- ~48% remain undetected

Threshold tuning could increase recall at the expense of precision, depending on retention campaign cost and business priorities.

In retention contexts, detecting churners (recall) is typically more valuable than maximising raw accuracy.
